```python

# SPOTIFY MUSIC ANALYSIS
# Data Exploration and Visualization

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
import io

sns.set_theme(style="whitegrid")

# 1. UPLOAD DATASET
print("Upload your Spotify CSV file:")
uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[filename]))

print("\nDataset loaded successfully!")
print("File:", filename)

# 2. DATA EXPLORATION
print("\n--- First 5 Rows ---")
print(df.head())

print("\n--- Dataset Shape ---")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\n--- Column Names ---")
print(df.columns.tolist())

print("\n--- Dataset Information ---")
df.info()

print("\n--- Missing Values ---")
print(df.isnull().sum())

print("\n--- Duplicate Rows ---")
print(df.duplicated().sum())

print("\n--- Statistical Summary ---")
print(df.describe(include="all").T)

# 3. DATA CLEANING
df = df.drop_duplicates().copy()

# Remove leading/trailing spaces from column names
df.columns = df.columns.str.strip()

# Standardize column names
df.columns = df.columns.str.lower().str.replace(" ", "_")

print("\nDataset shape after removing duplicates:", df.shape)

# 4. IDENTIFY USEFUL COLUMNS
print("\nAvailable columns:")
print(df.columns.tolist())

# Find columns even if names differ slightly
def find_column(possible_names):
    for name in possible_names:
        if name in df.columns:
            return name
    return None

pop_col = find_column(["popularity", "track_popularity"])
genre_col = find_column(["genre", "track_genre"])
artist_col = find_column(["artist", "artists", "artist_name"])
track_col = find_column(["track_name", "name", "track"])
dance_col = find_column(["danceability"])
energy_col = find_column(["energy"])
year_col = find_column(["release_year", "year"])
tempo_col = find_column(["tempo"])
duration_col = find_column(["duration_ms"])

# 5. POPULARITY DISTRIBUTION
if pop_col:
    plt.figure(figsize=(8, 5))
    sns.histplot(df[pop_col].dropna(), bins=20, kde=True)
    plt.title("Distribution of Song Popularity")
    plt.xlabel("Popularity")
    plt.ylabel("Number of Tracks")
    plt.tight_layout()
    plt.show()
else:
    print("\nPopularity column not found.")

# 6. TOP 10 POPULAR SONGS
if pop_col and track_col:
    cols = [track_col, pop_col]
    if artist_col:
        cols.insert(1, artist_col)

    top_songs = df.nlargest(10, pop_col)[cols]
    print("\n--- Top 10 Popular Songs ---")
    print(top_songs.to_string(index=False))

    plot_df = df.nlargest(10, pop_col).copy()
    labels = plot_df[track_col].fillna("Unknown").astype(str)

    plt.figure(figsize=(10, 6))
    plt.barh(labels, plot_df[pop_col])
    plt.title("Top 10 Most Popular Songs")
    plt.xlabel("Popularity")
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()

# 7. TOP 10 ARTISTS
if artist_col and pop_col:
    artist_pop = (
        df.groupby(artist_col)[pop_col]
        .mean()
        .sort_values(ascending=False)
        .head(10)
    )

    plt.figure(figsize=(10, 5))
    artist_pop.sort_values().plot(kind="barh")
    plt.title("Top 10 Artists by Average Track Popularity")
    plt.xlabel("Average Popularity")
    plt.tight_layout()
    plt.show()

# 8. GENRE DISTRIBUTION
if genre_col:
    genre_counts = df[genre_col].value_counts().head(10)

    plt.figure(figsize=(10, 5))
    genre_counts.sort_values().plot(kind="barh")
    plt.title("Top 10 Music Genres by Track Count")
    plt.xlabel("Number of Tracks")
    plt.tight_layout()
    plt.show()

# 9. DANCEABILITY VS ENERGY
if dance_col and energy_col:
    plt.figure(figsize=(8, 5))
    sns.scatterplot(
        data=df,
        x=dance_col,
        y=energy_col,
        alpha=0.5
    )
    plt.title("Danceability vs Energy")
    plt.xlabel("Danceability")
    plt.ylabel("Energy")
    plt.tight_layout()
    plt.show()

    print("\nDanceability-Energy correlation:",
          round(df[dance_col].corr(df[energy_col]), 3))

# 10. RELEASE YEAR TRENDS
if year_col:
    years = pd.to_numeric(df[year_col], errors="coerce")
    yearly_counts = years.dropna().astype(int).value_counts().sort_index()

    plt.figure(figsize=(10, 5))
    yearly_counts.plot(kind="line")
    plt.title("Number of Tracks by Release Year")
    plt.xlabel("Release Year")
    plt.ylabel("Number of Tracks")
    plt.tight_layout()
    plt.show()

# 11. TEMPO DISTRIBUTION
if tempo_col:
    plt.figure(figsize=(8, 5))
    sns.histplot(df[tempo_col].dropna(), bins=30, kde=True)
    plt.title("Distribution of Song Tempo")
    plt.xlabel("Tempo (BPM)")
    plt.ylabel("Number of Tracks")
    plt.tight_layout()
    plt.show()

# 12. DURATION ANALYSIS
if duration_col:
    duration_minutes = (
        pd.to_numeric(df[duration_col], errors="coerce") / 60000
    )

    plt.figure(figsize=(8, 5))
    sns.histplot(duration_minutes.dropna(), bins=30, kde=True)
    plt.title("Distribution of Song Duration")
    plt.xlabel("Duration (minutes)")
    plt.ylabel("Number of Tracks")
    plt.tight_layout()
    plt.show()

# 13. CORRELATION HEATMAP
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

if not correlation.empty:
    plt.figure(figsize=(12, 8))
    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        annot_kws={"size": 8}
    )
    plt.title("Correlation Between Numeric Features")
    plt.tight_layout()
    plt.show()

# 14. FINAL SUMMARY
print("\n========== PROJECT SUMMARY ==========")
print("Dataset:", filename)
print("Total tracks:", len(df))
print("Total columns:", len(df.columns))

if pop_col:
    print("Average popularity:",
          round(pd.to_numeric(df[pop_col], errors="coerce").mean(), 2))

if genre_col:
    print("Most common genre:",
          df[genre_col].mode().iloc[0]
          if not df[genre_col].mode().empty else "Unavailable")

print("\nSpotify Music Analysis completed!")
```
